```python
sales = pd.read_csv(r'D:\2nd_project\Retails Analytics\salesdata.csv')
```

```python
sales['order_id'].isnull().sum()
```

```python
sales['store_id'] = sales['store_id'].fillna('ONLINE')
```

```python
sales.head(50)
```

```python
sales["order_id"].duplicated()
```

```python
sales = sales.drop_duplicates(subset=["order_id"])
```

```python
sales['order_date'] = pd.to_datetime(
    sales['order_date'],
    dayfirst=True,
    errors='coerce'
)
```

```python
sales = sales.dropna(subset=['order_date'])
```

```python
sales['order_date'] = sales['order_date'].dt.strftime('%Y-%m-%d')
```

```python
sales
```

```python
products = pd.read_csv(r'D:\2nd_project\Retails Analytics\products.csv')
print(products.head())
```

```python
merge = sales.merge(
    products[['product_id', 'cost_price']],
    on='product_id',
    how='left'
)
```

```python
print(merge.head())
```

```python
merge['Gross_Profit'] = merge['total_amount'] - (merge['cost_price'] * merge['quantity'])
```

```python
print(merge[['product_id', 'total_amount', 'cost_price', 'quantity', 'Gross_Profit']].head())
```

```python
merge['cost_price'] = merge['cost_price'].fillna(0)
```

```python
merge['Profit_%'] = (merge['Gross_Profit'] / merge['total_amount']) * 100
```

```python
merge
```

```python
returns = pd.read_csv(r'D:\2nd_project\Retails Analytics\returns.csv')
```

```python
sales['Is_Returned'] = sales['order_id'].isin(returns['order_id']).astype(int)
```

```python
return_rate = sales['Is_Returned'].mean() * 100
print("Return Rate:", return_rate, "%")
```
