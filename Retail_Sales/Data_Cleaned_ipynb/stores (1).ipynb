```python
import pandas as pd
```

```python
import numpy as np
```

```python
store = pd.read_csv(r'D:\2nd_project\Retails Analytics\stores.csv')
```

```python
store.head(50)
```

```python
# Count duplicates in the column
duplicate_count = store.duplicated(subset=['store_id']).sum()
print(f"Number of duplicates in customer_id: {duplicate_count}")
```

```python
# Remove duplicates and keep the first occurrence
store = store.drop_duplicates(subset=['store_id'], keep='first')
```

```python
# % of null values per column
null_percentage = (store.isnull().sum() / len(store)) * 100
null_percentage = null_percentage.round(2)  # Round to 2 decimal places
print(null_percentage)
```

```python
# Capitalize first letters to keep it consistent
store['region'] = store['region'].str.strip().str.title()
store['city'] = store['city'].str.strip().str.title()
```

```python
store['operating_cost'] = pd.to_numeric(store['operating_cost'], errors='coerce')
```

```python
store.to_csv('sql_store.csv', index=False)
print("Returns data cleaned with age_group!")
```
