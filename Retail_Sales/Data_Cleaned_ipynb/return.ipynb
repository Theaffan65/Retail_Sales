```python
import numpy as np
```

```python
returns = pd.read_csv(r'D:\2nd_project\Retails Analytics\returns.csv')
```

```python
returns.head(50)
```

```python
# % of null values per column
null_percentage = (returns.isnull().sum() / len(returns)) * 100
null_percentage = null_percentage.round(2)  # Round to 2 decimal places
print(null_percentage)
```

```python
returns.dtypes
```

```python
returns['return_date'] = pd.to_datetime(
    returns['return_date'],
    dayfirst=True,     # agar aapka date format DD/MM/YYYY hai
    errors='coerce'    # invalid date ko NaT me convert kare
)
```

```python
invalid_dates = returns['return_date'].isnull().sum()
print(f"Invalid/NULL dates: {invalid_dates}")
```

```python
returns = returns.dropna(subset=['return_date'])
```

```python
returns['return_date'] = returns['return_date'].dt.strftime('%Y-%m-%d')
```
