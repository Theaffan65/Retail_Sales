```python
customer = pd.read_csv(r'D:\2nd_project\Retails Analytics\customers.csv')
```

```python
customer['signup_date'] = pd.to_datetime(customer['signup_date'], dayfirst=True).dt.strftime('%Y-%m-%d')
```

```python
customer
```

```python
customer= customer.dropna(subset=["age"])
```

```python
customer
```

```python
print(customer["customer_id"].duplicated())
```

```python
customer = customer.drop_duplicates(subset=["customer_id"])
```

```python
print(customer["customer_id"].duplicated())
```

```python
def create_age_group(age):
    if age <= 25: return 'Young'
    elif age <= 45: return 'Adult'
    elif age <= 60: return 'Middle-Aged'
    else: return 'Senior'

customer['age_group'] = customer['age'].apply(create_age_group)
```

```python
customer
```

```python
customer.to_csv('sql_customers.csv', index=False)
print("Customers data cleaned with age_group!")
```
